np中创建各种数组

```python
#使用list创建
a = np.array([2,3,4])
print(a) #array结果没有逗号

a = np.array([2,3,4],dtype=np.integer)
print(a.dtype) #设置数据类型

#定义多维array
a = np.array([[2,3,4],
              [3,4,5]])
#全0矩阵
a = np.zeros((3,4))
print(a)
#全1矩阵
a = np.ones((3,4),dtype=np.integer)
print(a)
#全空矩阵
a = np.empty((3,4))
print(a)

#数列
a = np.arange(10,20,2)
print(a)

a = np.arange(10,20,1).reshape((2,5))
print(a)

#线段
a = np.linspace(1,10,5)
print(a)
```

np的基础运算

```python
a = np.array([10,20,30,40])
b = np.arange(4)
print(a,b)

c = a - b
print(c)

c = b**2
print(c)

#三角函数
c = 10*np.sin(a)
print(c)
#判断array与某个值的大小
print(b<3)
print(b==3)
#运算矩阵
a = np.array([[1,1],
              [0,1]])
b = np.arange(4).reshape((2,2))
print(a)
print(b)
c = a*b #对应位置,逐个相乘
print(c)
c_dot = np.dot(a,b) #矩阵乘法
#上式等价于c_dot = a.dot(b)
print(c_dot)
#随机产生矩阵
a = np.random.random((2,4)) #(0,1)之间的数据
print(a)
#sum
print(np.sum(a))
print(np.min(a))
print(np.max(a))
#axis 可选某个维度下的最大值或最小值 axis=0是行,axis=1是列.
print(np.sum(a,axis=1))
```

Numpy的基础运算2\
索引

```python
a = np.arange(2,14).reshape(3,4)
print(a)

print(np.argmin(a))
print(np.argmax(a))

#计算数列的均值
print(np.mean(a))
print(a.mean())
print(np.average(a))
#中位数
print(np.median(a))
#数据累加
print(np.cumsum(a))
#数据累差
print(np.diff(a))
#找非0
print(np.nonzero(a)) #输出值的行数和列数
#sort
print(np.sort(a))
#计算矩阵相乘
print(np.transpose(a))
print(a.T)
#CLIP
print(np.clip(a,5,9)) #小于5就等于5,大于9就给9
print(np.mean(a,axis=0)) #对行求均值
```

Numpy的索引

```python
#一维
a = np.arange(3,15)
print(a)
print(a[3])
#二维
a = np.arange(3,15).reshape(3,4)
print(a)
print(a[2])
print(a[1,1])
print(a[1][1])
print(a[2,:])
print(a[:,1])
print(a[1,1:2])
#for 循环
#迭代行
for row in a:
  print(row)
#迭代列
for column in a.T:
  print(column)
#展平
print(a.flatten()) #返回一个被改变的flatten
for item in a.flat:
  print(item)
```

Numpy的array合并

```python
a = np.array([1,1,1])
b = np.array([2,2,2])

#上下合并
print(np.vstack((a,b))) #vertical stack
print(np.vstack((a,b)).shape)
print(a.shape)
#横向合并
print(np.hstack((a,b))) #horizontal stack
print(np.hstack((a,b)).shape)
#横向数列转纵向数列
#transpose做不到
print(a[np.newaxis,:])
print(a[np.newaxis,:].shape)
print(a[:,np.newaxis])
print(a[:,np.newaxis].shape)

#多项合并
a = np.array([1,1,1])
b = np.array([2,2,2])
c = np.concatenate((a,b,b,a),axis=0)
print(c)
```

Numpy的array的分割

```python
a = np.arange(12).reshape(3,4)
print(a)

print(np.split(a,2,axis=1))
#不等分割,split会报错
#print(np.split(a,3,axis=1))
#不等量分割
print(np.array_split(a,3,axis=1)) #加一个array就可以成功不等分割
#横向分割
print(np.vsplit(a,3))
#纵向分割
print(np.hsplit(a,2))
```

numpy的copy和deep copy

```python
a = np.arange(4)
b = a
c = a
d = b
a[0] = 11
a
b is a #a中数据改变也意味着b c d中的数据改变
d is a

#如何不要关联另外的变量
b = a.copy() #deep copy
a[3] = 44
print(b)
print(a)
b is a
```

Pandas选择数据

```python
import pandas as pd
import numpy as np
dates = pd.date_range('20160101',periods=6)
dates

df = pd.DataFrame(np.random.randn(6,4),index=dates,columns=['a','b','c','d'])
df

print(df['a'])
print(df.a)
print(df[0:3])
print(df['20160102':'20160104'])

#selected by label: loc
print(df.loc['20160102'])
print(df.loc[:,['a','b']])
print(df.loc['20160102',['a','b']])

#selected by position: iloc
print(df.iloc[3])
print(df.iloc[3,1])
print(df.iloc[3:5,1:3])
print(df.iloc[[1,2,4],[1,2]])

#mixed selection: ix
#print(df.ix[:3,['a','c']])
#新版本已经删除了ix

#Boolean indexing
print(df[df.a>0])
print(df[df>0])

```

pandas设置值

```python
dates = pd.date_range('20130101',periods=6)
df = pd.DataFrame(np.arange(24).reshape((6,4)),index=dates,columns=['A','B','C','D'])
df

df.iloc[2,2] = 1111
df.loc['20130101','B'] = 2222
df

df[df.A>4] = 0
print(df)

df['F'] = np.nan
df['E'] = pd.Series([1,2,3,4,5,6],index=pd.date_range('20130101',periods=6))
print(df)
```

pandas处理丢失数据

```python
dates = pd.date_range('20130101',periods=6)
df = pd.DataFrame(np.arange(24).reshape((6,4)),index=dates,columns=['A','B','C','D'])
df


df.iloc[1,2] = np.nan
df.iloc[2,3] = np.nan
df
#删除nan
print(df.dropna(axis=0,how='any')) #how={'all','any'}
print(df.dropna(axis=1,how='any'))

#填补为某个数字
print(df.fillna(value=0))

#检查是否为nan
print(df.isnull())
print(np.any(df.isnull()) == True)
```

pandas合并concat

```python
#concatenating
df1 = pd.DataFrame(np.ones((3,4))*0,columns=['a','b','c','d'])
df2 = pd.DataFrame(np.ones((3,4))*1,columns=['a','b','c','d'])
df3 = pd.DataFrame(np.ones((3,4))*2,columns=['a','b','c','d'])
print(df1)
print(df2)
print(df3)

#上下合并
res = pd.concat([df1,df2,df3],axis=0) #axis=0,竖向,axis=1,横向.
print(res)

res = pd.concat([df1,df2,df3],axis=0,ignore_index=True) #重排纵向
print(res)

#join,{'inner','outer'}
df1 = pd.DataFrame(np.ones((3,4))*0,columns=['a','b','c','d'],index=[1,2,3])
df2 = pd.DataFrame(np.ones((3,4))*1,columns=['b','c','d','e'],index=[2,3,4])
print(df1)
print(df2)
#有共同也有不同
res = pd.concat([df1,df2],join='outer') #outer是默认的
print(res)
res = pd.concat([df1,df2],join='inner',ignore_index=True)
print(res)
```
